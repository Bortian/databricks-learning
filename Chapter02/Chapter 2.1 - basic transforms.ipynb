```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *

spark = (SparkSession.builder
         .appName("basic-transforms")
         .master("spark://spark-master:7077")
         .config("spark.executor.memory", "512m")
         .getOrCreate()
        )
```

```python
df = (spark.read.format("json")
      .option("multiLine", "true")
      .load("../data/nobel_prizes.json")
     )
```

```python
df_transformed = (
    df.select("category",
              "overallMotivation",
              "year",
              "laureates",
              transform(col("laureates"),
                        lambda x: concat(x.firstname, lit(" "), x.surname))
              .alias("laureates_full_name")))

df_transformed.show()
```

```python
df_deduped = df.dropDuplicates(["category", "overallMotivation", "year"])
df_deduped.show()
```

```python
df_sorted = df.orderBy(["year", "category"])
df_sorted.show()
```

```python
df_renamed = df.withColumnRenamed("category", "Topic")
df_renamed.show()
```

```python
df_renamed = (
    df.selectExpr("category as Topic",
                  "year as Year",
                  "overallMotivation as Motivation"))

df_renamed.show()
```
